```python
#Import libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Abrir la imagen
bgr = cv2.imread('image.jpg')

# Mostrar el tamaño original de la imagen
print(f'Tamaño original de la imagen: {bgr.shape[1]}x{bgr.shape[0]}')

# Mostrar la imagen original en matplotlib
rgb = cv2.cvtColor(bgr, cv2.COLOR_BGR2RGB)
plt.imshow(rgb)
```

```python
# Mostrar la imagen en escala de grises
gray = cv2.cvtColor(bgr, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.axis('off')
```

```python
#Mostrar los componentes de color de la imagen original
b, g, r = cv2.split(bgr)        
plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(b, cmap='gray')
plt.axis('off')
plt.title('Canal Azul')
plt.subplot(1,3,2)
plt.imshow(g, cmap='gray')
plt.axis('off')
plt.title('Canal Verde')
plt.subplot(1,3,3)
plt.imshow(r, cmap='gray')
```

```python
#Mostrar los componentes HLKS de la imagen 
hls = cv2.cvtColor(bgr, cv2.COLOR_BGR2HLS)
h, l, s = cv2.split(hls)

plt.figure(figsize=(12,4))
plt.subplot(1,3,1)
plt.imshow(h, cmap='gray')


plt.axis('off')
plt.title('Canal Matiz (Hue)')
plt.subplot(1,3,2)
plt.imshow(l, cmap='gray')
plt.axis('off')
plt.title('Canal Luminosidad (Lightness)')
plt.subplot(1,3,3)
plt.imshow(s, cmap='gray')
plt.axis('off')
plt.title('Canal Saturación (Saturation)')
plt.show()
```

```python
#Ajuste el brillo y contraste de la imagen
alpha = 1.5  # Contraste control (1.0-3.0)
beta = 50    # Brillo control (0-100)
adjusted = cv2.convertScaleAbs(bgr, alpha=alpha, beta=beta)
plt.imshow(cv2.cvtColor(adjusted, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Imagen con brillo y contraste ajustados')
plt.show()  
```

```python
#Aplicar filtro Gaussiano
blurred = cv2.GaussianBlur(bgr, (15, 15), 0)
plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.title('Imagen con filtro Gaussiano')
plt.show()
```
